```python
from __future__ import annotations
from typing import ClassVar, Final

class MyClass:
    """
    Short summary.

    Longer description explaining the purpose of the class,
    invariants, and any important implementation details.
    """

    # -------------------------------------------------------------------------
    # Class constants
    # -------------------------------------------------------------------------

    DEFAULT_NAME: Final[str] = "Unnamed"

    # -------------------------------------------------------------------------
    # Class attributes (shared by all instances)
    # -------------------------------------------------------------------------

    all: ClassVar[list[MyClass]] = []
    instance_count: ClassVar[int] = 0

    # -------------------------------------------------------------------------
    # Constructor
    # -------------------------------------------------------------------------

    def __init__(
        self,
        name: str,
        value: float,
        enabled: bool = True,
    ) -> None:
        """
        Initialize a new MyClass instance.

        Args:
            name: Object name.
            value: Numeric value.
            enabled: Whether the object is active.
        """

        if value < 0:
            raise ValueError("value must be non-negative")

        self.name = name
        self.value = value
        self.enabled = enabled

        # Internal attribute
        self._cached_result: float | None = None

        # "Private" attribute (name mangling)
        self.__secret = "hidden"

        type(self).instance_count += 1
        type(self).all.append(self)

    # -------------------------------------------------------------------------
    # Properties
    # -------------------------------------------------------------------------

    @property
    def is_enabled(self) -> bool:
        """Return whether the object is enabled."""
        return self.enabled

    @property
    def cached_result(self) -> float | None:
        """Read-only cached value."""
        return self._cached_result

    # -------------------------------------------------------------------------
    # Public methods
    # -------------------------------------------------------------------------

    def calculate(self) -> float:
        """Calculate and return a value."""
        return self.value * 2

    # -------------------------------------------------------------------------
    # Protected helper methods
    # -------------------------------------------------------------------------

    def _reset_cache(self) -> None:
        """Reset cached data."""
        self._cached_result = None

    # -------------------------------------------------------------------------
    # Class methods
    # -------------------------------------------------------------------------

    @classmethod
    def from_string(cls, text: str) -> MyClass:
        """Alternative constructor."""
        name, value = text.split(",")
        return cls(name, float(value))

    # -------------------------------------------------------------------------
    # Static methods
    # -------------------------------------------------------------------------

    @staticmethod
    def is_valid_name(name: str) -> bool:
        """Validate a name."""
        return bool(name.strip())

    # -------------------------------------------------------------------------
    # Magic methods
    # -------------------------------------------------------------------------

    def __repr__(self) -> str:
        return (
            f"{type(self).__name__}("
            f"name={self.name!r}, "
            f"value={self.value!r}, "
            f"enabled={self.enabled!r})"
        )

    def __str__(self) -> str:
        return self.name

    def __eq__(self, other: object) -> bool:
        if not isinstance(other, MyClass):
            return NotImplemented
        return (self.name, self.value) == (other.name, other.value)

    def __lt__(self, other: MyClass) -> bool:
        return self.value < other.value
```
